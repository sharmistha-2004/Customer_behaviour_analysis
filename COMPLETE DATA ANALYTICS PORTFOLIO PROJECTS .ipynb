# # to check the missing values or null values in data

```python
df.isnull().sum()
```

```python
''' 
# we will handle missing rating or null values using mean or median 

# generally we should choose median over mean because mean can get affected by outliers and that skew the distribution of 
data

# and why median ? because median is robust to outliers

# now a common beginner will do here is just fill all the missing rating with over all median of entire column and moove on
and yes that technically works but its not the best practice, 

why ? because rating can vary a lot depending on the category like clothing can have different review rating than footware 
category similarly 

# now if we just take the overall median and fill it everywhere, we are basically forcing clothing items to get the same 
review ratings as footware items and that can introduce bias into our data set instead what we will do is a little bit smarter
than that

# we will impute the missing values using the median review rating within each category

# we will find out the median review rating within each category and will replace that with the missing values of the category

# if the clothing item review rating is missing , its gets replaced with median review rating of the clothing items 
not footwear item lets write the code
```

# we will group it by category, we will use fillna for that to fill it with median
# values within that category.

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
# now check the missing and null values are left, we have imputed all the null values in the review rating column to clean it 
# this is the right way to clean the data
```

```python
df.isnull().sum()
```

```python
# data formatting is needed because here available upper case, lower case, speaces we will use snake casing(all lower case)
# and all specaes are replaced by _(underscore)
```

```python
# lower case all the columns name
# replace all the spaces with _ (underscore)
# short the long name with suitable name
```

```python
df.columns = df.columns.str.lower() 
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
# lets do some future engineering, add a new column age_group
```

```python
labels = ['Young Adult','Adult','Middle-aged','Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels)
```

```python
df[['age','age_group']].head(10)
```

```python
# lets do some future engineering, create column purchase_frequency_days
```

```python
frequency_mapping ={
    
    'Fortnightly': 14,
    'Weekly':7,
    'Monthly':30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90    
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```
